## 2. Grain & Candidate Key Verification (`order_id`)
Memeriksa kolom kunci `order_id` agar bebas dari null dan duplikasi.
* **Penentuan Grain:** 1 baris merepresentasikan 1 pesanan (*single order grain*).
* **Penentuan Candidate Key:** Kolom `order_id` ditetapkan sebagai candidate key utama di kedua dataset.
* **Pemeriksaan Integritas:** Memverifikasi keunikan (0 duplikat) dan kelengkapan (0 null) pada kolom `order_id`.

```python
# 1. Definisi Grain & Candidate Key
print("--- Penentuan Grain & Candidate Key ---")
print("Target Grain        : 1 baris = 1 pesanan (order)")
print("Candidate Key       : order_id")

# 2. Verifikasi Keunikan (Duplikat) & Kelengkapan (Null)
pm2_total = len(df_pm2)
pm2_unique = df_pm2["order_id"].nunique()
pm2_null = df_pm2["order_id"].isnull().sum()

rev_total = len(df_reviews)
rev_unique = df_reviews["order_id"].nunique()
rev_null = df_reviews["order_id"].isnull().sum()

r"""
3. Cetak Hasil Pembuktian Kriteria Langkah 2
"""
print("\n--- Verifikasi Integritas order_id ---")
print(f"PM2     -> Total: {pm2_total} | Unique: {pm2_unique} | Null: {pm2_null} | Duplikat: {pm2_total - pm2_unique}")
print(f"Reviews -> Total: {rev_total} | Unique: {rev_unique} | Null: {rev_null} | Duplikat: {rev_total - rev_unique}")

# Validasi Otomatis Status Candidate Key
is_pm2_valid = (pm2_total == pm2_unique) and (pm2_null == 0)
is_rev_valid = (rev_total == rev_unique) and (rev_null == 0)

print(f"\nApakah order_id valid sebagai Candidate Key? : {'YA (VALID)' if (is_pm2_valid and is_rev_valid) else 'TIDAK'}")
```

## 3. Relational Key Matching (PM2 vs Reviews)
Mencocokkan daftar `order_id` pada dataset Reviews terhadap `order_id` di final PM2 untuk memastikan tidak ada relasi ulasan yang hilang/mismatch sebelum dilakukan integrasi.

```python
# Himpunan order_id kedua dataset
set_pm2_orders = set(df_pm2["order_id"])
set_review_orders = set(df_reviews["order_id"])

# Cek matched dan unmatched
unmatched_reviews = set_review_orders - set_pm2_orders
matched_reviews = set_review_orders.intersection(set_pm2_orders)

print("=== HASIL PENCOCOKAN RELASI ORDER_ID ===")
print("Total order_id unik di PM2          :", len(set_pm2_orders))
print("Total order_id unik di Reviews      :", len(set_review_orders))
print("Matched Reviews (cocok dengan PM2)  :", len(matched_reviews))
print("Unmatched Reviews (tidak ada di PM2):", len(unmatched_reviews))

if len(unmatched_reviews) == 0:
    print("\nSTATUS LANGKAH 3: PASS (Relasi aman untuk lanjut ke LEFT JOIN)")
else:
    print(f"\nSTATUS LANGKAH 3: FAILED ({len(unmatched_reviews)} ulasan tidak cocok)")
```

# Tahap 02: Integrasi PM2 + Reviews (Join Reviews)
*Referensi: Panduan Capstone Hal. 9 & 11 (Penggabungan Tanpa Fan-Out)*

* **Tujuan Tahap:** Menggabungkan dataset PM2 dengan ulasan `SB-P18` menggunakan `LEFT JOIN` pada `order_id` tanpa mengubah *grain* (tetap 1 baris per pesanan) serta membuktikan 6 kriteria audit validasi.
* **Output yang Diharapkan:** Dataset kerja terintegrasi (`df_integrated`) sebanyak 324 baris dan tabel bukti (*evidence*) 6 pemeriksaan wajib berstatus `PASS`.

## 4. Eksekusi LEFT JOIN
Melakukan LEFT JOIN dari data transaksi PM2 ke Reviews berbasis candidate key `order_id` setelah seluruh risiko mismatch dan duplicate dipastikan aman (bebas fan-out).

```python
# Langkah 4 PDF: Eksekusi LEFT JOIN
df_integrated = df_pm2.merge(df_reviews, on="order_id", how="left")

# Verifikasi awal tidak ada fan-out
print("=== HASIL EKSEKUSI LEFT JOIN ===")
print("Jumlah baris PM2 (sebelum join)         :", len(df_pm2))
print("Jumlah baris terintegrasi (setelah join) :", len(df_integrated))
print("Jumlah kolom setelah join               :", df_integrated.shape[1])

# Cek fan-out
if len(df_integrated) == len(df_pm2):
    print("\nSTATUS LANGKAH 4: PASS (Baris tetap 324, terbukti TIDAK ADA fan-out)")
else:
    print(f"\nSTATUS LANGKAH 4: FAILED (Terjadi fan-out! Baris membengkak jadi {len(df_integrated)})")
```

## 5. Fitur Review & Penanganan Tiga Kondisi
Menambahkan fitur `has_review` dan `review_status`, serta memisahkan 3 kondisi perlakuan ulasan (dengan teks, tanpa teks, dan No Review) sesuai kontrak data.

```python
# 1. Tambahkan flag has_review dan label review_status
df_integrated["has_review"] = df_integrated["rating"].notnull().astype(int)
df_integrated["review_status"] = df_integrated["rating"].apply(
    lambda x: "Reviewed" if pd.notnull(x) else "No Review"
)

# 2. Logika pemisahan tiga kondisi ulasan sesuai tabel PDF
has_text = df_integrated["review_text"].notnull() & (df_integrated["review_text"].astype(str).str.strip() != "")
cond_with_text = (df_integrated["has_review"] == 1) & has_text
cond_no_text = (df_integrated["has_review"] == 1) & (~has_text)
cond_no_review = df_integrated["has_review"] == 0

print("=== DISTRIBUSI TIGA KONDISI REVIEW ===")
print("1. Ada review dengan teks :", cond_with_text.sum())
print("2. Ada review tanpa teks  :", cond_no_text.sum())
print("3. Tidak ada review       :", cond_no_review.sum())
print("-" * 38)
print("Total pesanan terverifikasi:", len(df_integrated))

# Validasi kelengkapan total baris
if (cond_with_text.sum() + cond_no_text.sum() + cond_no_review.sum()) == len(df_integrated):
    print("\nSTATUS LANGKAH 5: PASS (Tiga kondisi terklasifikasi 100% konsisten)")
else:
    print("\nSTATUS LANGKAH 5: FAILED (Ada baris ulasan tidak terklasifikasi)")
```

## 6. Simpan Working Dataset & Validasi Evidence
Menyimpan working dataset (`df_integrated`) sebagai file CSV perantara untuk analisis bisnis di Tahap 03, serta memverifikasi 6 kriteria audit kualitas data tanpa fan-out.

```python
from pathlib import Path

# 1. Simpan working dataset ke folder processed
output_dir = base_dir / "data" / "processed" / "capstone_swiftbite"
output_dir.mkdir(parents=True, exist_ok=True)

working_file = output_dir / "swiftbite_working_integrated.csv"
df_integrated.to_csv(working_file, index=False)
print("File tersimpan di:", working_file.name)

# 2. Audit 6 Pemeriksaan Wajib (Halaman 11 PDF)
pm2_rows = len(df_pm2)
capstone_rows = len(df_integrated)
unique_order_id = df_integrated["order_id"].nunique()
duplicate_order_id = df_integrated["order_id"].duplicated().sum()
unmatched_reviews = len(set(df_reviews["order_id"]) - set(df_pm2["order_id"]))
matched_reviews = df_integrated["has_review"].sum()
missing_reviews = (df_integrated["has_review"] == 0).sum()

# Monetary Check (successful_payment_amount & refunded_payment_amount)
diff_successful = abs(df_pm2["successful_payment_amount"].sum() - df_integrated["successful_payment_amount"].sum())
diff_refunded = abs(df_pm2["refunded_payment_amount"].sum() - df_integrated["refunded_payment_amount"].sum())
monetary_status = "0.0 (Stabil/Cocok)" if (diff_successful == 0 and diff_refunded == 0) else f"Diff Succ: {diff_successful}, Diff Ref: {diff_refunded}"

# Safe Data Check (Pemeriksaan PII: ditambah customer_id, address, salt, secret)
forbidden_pii = [
    "full_name", "customer_name", "email", "phone_number", "phone",
    "raw_customer_id", "controlled_salt", "customer_id", "address", "salt", "secret"
]
detected_pii = [c for c in df_integrated.columns if c.lower() in forbidden_pii]

print("\n" + "=" * 45)
print("EVIDENCE FORMAT SETELAH JOIN REVIEWS")
print("=" * 45)
print(f"Variant ID              : {VARIANT_ID}")
print(f"PM2 rows                : {pm2_rows}")
print(f"Capstone rows           : {capstone_rows}")
print(f"Unique order_id         : {unique_order_id}")
print(f"Matched review order_id : {matched_reviews}")
print(f"Missing review (No Rev) : {missing_reviews}")
print(f"Unmatched Reviews       : {unmatched_reviews}")
print(f"Duplicate order_id      : {duplicate_order_id}")
print(f"Monetary diff           : {monetary_status}")
print(f"Direct PII field count  : {len(detected_pii)}")

# PASS hanya jika seluruh assert lolos
assert capstone_rows == pm2_rows, "Gagal: Jumlah baris berubah setelah join!"
assert unique_order_id == pm2_rows, "Gagal: order_id tidak unik!"
assert duplicate_order_id == 0, "Gagal: Terdapat duplikat order_id!"
assert unmatched_reviews == 0, "Gagal: Terdapat unmatched reviews!"
assert diff_successful == 0, "Gagal: Nilai successful_payment_amount tidak cocok!"
assert diff_refunded == 0, "Gagal: Nilai refunded_payment_amount tidak cocok!"
assert len(detected_pii) == 0, f"Gagal: Terdeteksi Direct PII ({detected_pii})!"

print("Final judgment          : PASS")
```

> * **Tiga Kondisi Ulasan:** 175 ada teks, 16 tanpa teks, dan 133 pesanan berstatus *No Review*.
> * **Monetary & Safe Data:** Selisih nominal Rp 0,00 (Stabil/Cocok) dan 0 kolom Direct PII terdeteksi (Tidak terjadi Direct PII/Personally Identifiable Information).
> * **Kesimpulan & Status Gateway:** `PASS` — Working dataset (`swiftbite_working_integrated.csv`) siap dipakai untuk hitung 4 KPI di Tahap 03.

## 9. Pemeriksaan Review Text & Deteksi Ambigu

*(Referensi: Panduan Capstone Hal. 15 - Langkah 1: Klasifikasi & Validasi Sentimen)*

Memeriksa kolom `review_text`, memisahkan ulasan kosong (`No Text`), serta mengidentifikasi ulasan yang berpotensi ambigu atau bertolak belakang dengan rating bintang.

```python
# 1. PROFILING TEKS ULASAN & DETEKSI POTENSI AMBIGU

# Ambil order yang memiliki record ulasan (has_review == True)
df_reviews = df_integrated[df_integrated["has_review"] == True].copy()

# 1. Identifikasi teks kosong (No Text) vs ada teks
kosong_mask = df_reviews["review_text"].isna() | (df_reviews["review_text"].astype(str).str.strip() == "")
total_kosong = int(kosong_mask.sum())
total_ada_teks = int((~kosong_mask).sum())

# 2. Deteksi teks yang berpotensi ambigu / kontradiktif dengan rating bintang:
# Kasus A: Rating tinggi (4 atau 5) tapi mengandung kata komplain
kata_negatif = ["lama", "telat", "dingin", "tumpah", "asin", "kecewa", "kurang", "rusak", "bocor", "salah"]
pattern_negatif = "|".join(kata_negatif)

ambigu_rating_tinggi = df_reviews[
    (~kosong_mask) & 
    (df_reviews["rating"].isin([4, 5])) & 
    (df_reviews["review_text"].astype(str).str.lower().str.contains(pattern_negatif, na=False))
]

# Kasus B: Rating rendah (1 atau 2) tapi mengandung kata pujian
kata_positif = ["enak", "cepat", "mantap", "puas", "bagus", "lezat", "rapi"]
pattern_positif = "|".join(kata_positif)

ambigu_rating_rendah = df_reviews[
    (~kosong_mask) & 
    (df_reviews["rating"].isin([1, 2])) & 
    (df_reviews["review_text"].astype(str).str.lower().str.contains(pattern_positif, na=False))
]

print(f"Total Record Ulasan      : {len(df_reviews)}")
print(f"- Ulasan Tanpa Teks      : {total_kosong} (Kategori: No Text)")
print(f"- Ulasan Berisi Teks     : {total_ada_teks}")
print("-" * 60)
print(f"Potensi Anomali Kasus A (Rating 4-5 tapi teks bernada keluhan): {len(ambigu_rating_tinggi)} data")
print(f"Potensi Anomali Kasus B (Rating 1-2 tapi teks bernada pujian) : {len(ambigu_rating_rendah)} data")
print("=" * 60)

# Tampilkan 5 contoh ulasan berpotensi anomali jika ditemukan
sampel_ambigu = pd.concat([ambigu_rating_tinggi, ambigu_rating_rendah]).head(5)
if len(sampel_ambigu) > 0:
    print("Contoh Teks yang Perlu Verifikasi Makna (Abaikan Rating Bintang):")
    for idx, row in sampel_ambigu.iterrows():
        print(f"[{row['order_id']}] Rating: {row['rating']} | Teks: \"{row['review_text']}\"")
else:
    print("Tidak ditemukan kontradiksi langsung berbasis kata kunci.")
```

### 11. Eksekusi Klasifikasi Sentimen Ulasan

*(Referensi: Panduan Capstone Hal. 15 - Langkah 3: Jalankan klasifikasi review)*

Menjalankan Klasifikasi Ulasan menggunakan total record ulasan hasil NO.9 df_reviews sebesar 191 Ulasan


```python
import re

kata_positif = [
    "enak", "lezat", "mantap", "puas", "bagus", "cepat", "ramah", "rapi",
    "hangat", "segar", "fresh", "suka", "pas", "rekomendasi", "terbaik", "juara",
    "sopan", "lancar", "memuaskan", "banyak", "baik", "worth"
]
kata_negatif = [
    "lama", "telat", "terlambat", "lambat", "dingin", "tumpah", "asin",
    "kecewa", "kurang", "rusak", "bocor", "salah", "basi", "buruk", "parah", "hambar",
    "sedikit", "bau", "keasinan", "berantakan", "sulit", "susah", "mahal"
]
kata_netral = [
    "standar", "biasa", "cukup", "lumayan", "sesuai", "rata-rata", "sampai", "seimbang"
]

FRASA_NEG = [
    r"kurang rapi", r"tidak terlalu hangat", r"kurang baik",
    r"kurang fresh", r"kurang enak", r"agak lama", r"agak mahal",
    r"tidak rapi", r"tidak hangat", r"tidak enak"
]
FRASA_NEGASI_POS = [
    r"tidak ada yang (?:tumpah|bocor|rusak|salah)",
    r"tidak (?:tumpah|bocor|rusak|salah)",
    r"tanpa (?:tumpah|bocor|rusak)"
]
KONTRAS = r"\b(?:tetapi|tapi|namun)\b"

def hitung_kata(daftar, teks):
    return sum(1 for k in daftar if re.search(r"\b" + re.escape(k), teks))

def proses_klasifikasi(row):
    text = row["review_text"]
    if pd.isna(text) or str(text).strip() == "":
        return "No Text"
    t = str(text).lower().strip()

    skor_pos = skor_neg = 0
    for p in FRASA_NEGASI_POS:
        n = len(re.findall(p, t))
        skor_pos += n
        t = re.sub(p, " ", t)

    for p in FRASA_NEG:
        n = len(re.findall(p, t))
        skor_neg += n
        t = re.sub(p, " ", t)

    skor_pos += hitung_kata(kata_positif, t)
    skor_neg += hitung_kata(kata_negatif, t)
    skor_net  = hitung_kata(kata_netral, t)

    # Prioritas mitigasi: ada keluhan -> Negative (campuran tetap Negative)
    if skor_neg > 0 and (skor_neg >= skor_pos or re.search(KONTRAS, t) or skor_neg > 0 and skor_pos == 0):
        return "Negative"
    if skor_neg > 0 and skor_pos > skor_neg:
        return "Negative"
    if skor_pos > 0:
        return "Positive"
    if skor_net > 0:
        return "Neutral"
    return "Unknown"

# Jalankan klasifikasi langsung ke df_reviews dari nomor 9
df_reviews["sentiment_label"] = df_reviews.apply(proses_klasifikasi, axis=1)

print("Status: Berhasil dijalankan dengan leksikon baru & penanganan negasi/kontras!")
print("Distribusi Sentimen pada 191 Ulasan:")
print(df_reviews["sentiment_label"].value_counts(dropna=False))

# Regression Asserts
assert (df_reviews[df_reviews["review_text"].str.contains("tidak ada yang tumpah", na=False)]["sentiment_label"] == "Positive").all(), "Gagal Assert Pujian Negasi"
assert (df_reviews[df_reviews["review_text"].str.contains("kurang rapi|tidak terlalu hangat", na=False)]["sentiment_label"] == "Negative").all(), "Gagal Assert Campuran Negatif"
assert (df_reviews["sentiment_label"] == "Unknown").sum() == 0, "Gagal Assert Unknown == 0"
assert (df_reviews["sentiment_label"] == "No Text").sum() == 16, "Gagal Assert No Text == 16"
assert len(df_reviews) == 191, "Gagal Assert Total Reviews == 191"
print("Regression Asserts: ALL PASS")
```

### 15. Kode Pembuktian Kuantitatif Temuan Bisnis (BQ 1 s.d. BQ 4)
*Blok kode ini mengeksekusi kalkulasi agregat untuk menghasilkan angka bukti kuantitatif (evidence) yang dicantumkan pada Matriks Temuan Bisnis Tahap 05 di atas:*
1. **5A (BQ 1):** Tren Bulanan Pesanan & Kontraksi Omzet (-32,0%).
2. **5B (BQ 2):** Breakdown Omzet & Keterlambatan 5 Zona (West Zone #1 Omzet & Molor +18,5 mnt).
3. **5C (BQ 3):** Peringkat 22 Resto Mitra (Partner 14 Telat 69,2%) & SLA Kategori Menu Western (38,9%).
4. **5D (BQ 4):** Komparasi Kepuasan On-Time vs Late (Rating 3,97 vs 2,92 ⭐) & Rating-Teks Tidak Sejalan (52,3% Komplain di ⭐ 4-5, dan 14 Teks Positif di ⭐ 1-2) (52,3%).


```python
import pandas as pd
from pathlib import Path

# 1. Load Analysis-Ready Dataset
base_dir = Path.cwd()
if base_dir.name == "notebooks":
    base_dir = base_dir.parent
csv_path = base_dir / "data" / "processed" / "capstone_swiftbite" / "swiftbite_capstone_analysis_ready.csv"
df = pd.read_csv(csv_path)

# Fitur pembantu
df["bulan"] = pd.to_datetime(df["order_timestamp"]).dt.to_period("M")
df["net"] = df["successful_payment_amount"] - df["refunded_payment_amount"]
df["valid"] = ((df["delivery_status"].astype(str).str.lower() == "delivered")
               & df["actual_delivery_minutes"].notnull()
               & df["promised_delivery_minutes"].notnull())
df["late"] = df["valid"] & (df["actual_delivery_minutes"] > df["promised_delivery_minutes"])
df["delay_mins"] = (df["actual_delivery_minutes"] - df["promised_delivery_minutes"]).clip(lower=0)

print("=" * 75)
print("BUKTI KUANTITATIF TAHAP 05: 4 PILAR TEMUAN BISNIS (BQ 1 - BQ 4)")
print("=" * 75)

# --- 5A (BQ 1): TREN BULANAN ---
print("\n--- 5A. BUKTI BQ 1: TREN BULANAN (KONTRAKSI OMZET -32,0%) ---")
g_bln = df.groupby("bulan").agg(
    orders=("order_id", "nunique"),
    net_rp=("net", "sum"),
    valid=("valid", "sum"),
    late=("late", "sum")
).reset_index()
g_bln["late_pct"] = (g_bln["late"] / g_bln["valid"] * 100).round(2)
for _, r in g_bln.iterrows():
    print(f"Bulan {r['bulan']} | Orders: {r['orders']:>3} | Net: Rp {int(r['net_rp']):>10,} ({r['net_rp']/1e6:.2f} Jt) | Late: {int(r['late']):>2}/{int(r['valid']):>2} ({r['late_pct']:>5}%)")

# --- 5B (BQ 2): PERFORMA 5 ZONA WILAYAH & DURASI MOLOR RESMI ---
print("\n--- 5B. BUKTI BQ 2: PERFORMA 5 ZONA WILAYAH & DURASI KETERLAMBATAN ---")
g_zona = df.groupby("customer_zone").agg(
    orders=("order_id", "nunique"),
    net_rp=("net", "sum"),
    valid=("valid", "sum"),
    late=("late", "sum")
).reset_index()
g_zona["omzet_pct"] = (g_zona["net_rp"] / df["net"].sum() * 100).round(2)
g_zona["late_pct"] = (g_zona["late"] / g_zona["valid"] * 100).round(2)

# Definisi baku durasi molor: rata-rata delay pada pesanan terlambat
avg_delays = []
for z in g_zona["customer_zone"]:
    subset = df[(df["customer_zone"] == z) & (df["late"] == True)]
    diffs = subset["actual_delivery_minutes"] - subset["promised_delivery_minutes"]
    avg_delays.append(round(diffs.mean(), 1))
g_zona["avg_molor"] = avg_delays
g_zona = g_zona.sort_values("net_rp", ascending=False)
for _, r in g_zona.iterrows():
    print(f"Zona {r['customer_zone']:<7} | Omzet: Rp {int(r['net_rp']):>10,} ({r['omzet_pct']:>5}%) | Late: {int(r['late']):>2}/{int(r['valid']):>2} ({r['late_pct']:>5}%) | Avg Molor: +{r['avg_molor']} mnt")

all_late = df[df["late"] == True]
diff_all = all_late["actual_delivery_minutes"] - all_late["promised_delivery_minutes"]
print(f"Rata-rata Molor Keseluruhan Platform: +{diff_all.mean():.1f} menit")

# --- 5C (BQ 3): TOP RESTO TELAT & 6 KATEGORI MENU ---
print("\n--- 5C. BUKTI BQ 3: MITRA PALING TELAT (PARTNER 14) & 6 KATEGORI MENU ---")
g_resto = df.groupby("restaurant_name").agg(orders=("order_id", "nunique"), valid=("valid", "sum"), late=("late", "sum")).reset_index()
g_resto["late_pct"] = (g_resto["late"] / g_resto["valid"] * 100).round(2)
g_resto = g_resto.sort_values(["late_pct", "late"], ascending=False)
for i, (_, r) in enumerate(g_resto.head(5).iterrows(), 1):
    print(f"Rank {i} | {r['restaurant_name']:<22} | Valid: {int(r['valid']):>2} | Late: {int(r['late']):>2} ({r['late_pct']:>5}%)")

g_menu = df.groupby("cuisine_type").agg(
    orders=("order_id", "nunique"),
    valid=("valid", "sum"),
    late=("late", "sum"),
    avg_rating=("rating", "mean")
).reset_index()
g_menu["late_pct"] = (g_menu["late"] / g_menu["valid"] * 100).round(2)
g_menu = g_menu.sort_values("late_pct", ascending=False)
print("SLA Resmi 6 Kategori Menu Makanan:")
for _, r in g_menu.iterrows():
    print(f" - Menu {r['cuisine_type']:<16} | Orders: {r['orders']:>2} | Valid: {int(r['valid']):>2} | Late: {int(r['late']):>2} ({r['late_pct']:>5}%) | Avg Rating: {r['avg_rating']:.2f}")

# --- 5D (BQ 4): DAMPAK KEPUASAN (ON-TIME VS LATE) & RATING-TEKS TIDAK SEJALAN ---
print("\n--- 5D. BUKTI BQ 4: DAMPAK KEPUASAN (ON-TIME VS LATE) & RATING-TEKS TIDAK SEJALAN ---")
df_deliv = df[df["valid"] == True]
df_rev = df_deliv[df_deliv["has_review"] == 1]
ot_rev = df_rev[df_rev["late"] == False]
lt_rev = df_rev[df_rev["late"] == True]

avg_ot = ot_rev["rating"].mean()
avg_lt = lt_rev["rating"].mean()
pct_pos_ot = (ot_rev["sentiment_label"] == "Positive").sum() / len(ot_rev) * 100
pct_neg_ot = (ot_rev["sentiment_label"] == "Negative").sum() / len(ot_rev) * 100
pct_pos_lt = (lt_rev["sentiment_label"] == "Positive").sum() / len(lt_rev) * 100
pct_neg_lt = (lt_rev["sentiment_label"] == "Negative").sum() / len(lt_rev) * 100

print(f"Saat On-Time ({len(ot_rev)} ulasan) : Avg Rating = {avg_ot:.2f} ⭐ | Positif = {pct_pos_ot:.1f}% ({int((ot_rev['sentiment_label'] == 'Positive').sum())}/{len(ot_rev)}) | Negatif = {pct_neg_ot:.1f}% ({int((ot_rev['sentiment_label'] == 'Negative').sum())}/{len(ot_rev)})")
print(f"Saat Late    ({len(lt_rev)} ulasan)  : Avg Rating = {avg_lt:.2f} ⭐ | Positif = {pct_pos_lt:.1f}% ({int((lt_rev['sentiment_label'] == 'Positive').sum())}/{len(lt_rev)}) | Negatif = {pct_neg_lt:.1f}% ({int((lt_rev['sentiment_label'] == 'Negative').sum())}/{len(lt_rev)}) (Drop {avg_lt - avg_ot:.2f} ⭐)")

all_neg = df[df["sentiment_label"] == "Negative"]
neg_tinggi = all_neg["rating"].isin([4, 5]).sum()
pct_bias = neg_tinggi / len(all_neg) * 100
all_pos = df[df["sentiment_label"] == "Positive"]
pos_rendah = all_pos["rating"].isin([1, 2]).sum()
pct_pos_rendah = pos_rendah / len(all_pos) * 100
print(f"Rating-Teks Tidak Sejalan: {neg_tinggi} dari {len(all_neg)} ulasan negatif ({pct_bias:.1f}%) berada di rating 4-5!")
print(f"Sebaliknya: {pos_rendah} dari {len(all_pos)} ulasan positif ({pct_pos_rendah:.1f}%) berada di rating 1-2!")
print("=" * 75)
```

```python
from pathlib import Path
import pandas as pd

base_dir = Path.cwd()
if base_dir.name == "notebooks":
    base_dir = base_dir.parent
csv_path = base_dir / "data" / "processed" / "capstone_swiftbite" / "swiftbite_capstone_analysis_ready.csv"

df = pd.read_csv(csv_path)
df["bulan"] = pd.to_datetime(df["order_timestamp"]).dt.to_period("M")

def kpi(s):
    n = len(s)
    net = int(s["successful_payment_amount"].sum() - s["refunded_payment_amount"].sum()) if n > 0 else 0
    rev = int(s["has_review"].sum()) if n > 0 else 0
    d = s[(s["delivery_status"].astype(str).str.lower() == "delivered")
          & s["actual_delivery_minutes"].notnull()
          & s["promised_delivery_minutes"].notnull()]
    late = int((d["actual_delivery_minutes"] > d["promised_delivery_minutes"]).sum())
    return n, net, rev, len(d), late

P07 = df["restaurant_name"] == "SwiftBite Partner 07"
tests = {
    "T-01 Tanpa filter": pd.Series(True, index=df.index),
    "T-02 Jan": df["bulan"] == "2026-01",
    "T-03 Western": df["cuisine_type"] == "Western",
    "T-04 P07+Asian": P07 & (df["cuisine_type"] == "Asian"),
    "T-05 Jan+Western": (df["bulan"] == "2026-01") & (df["cuisine_type"] == "Western"),
    "T-06 P01+Western": (df["restaurant_name"] == "SwiftBite Partner 01") & (df["cuisine_type"] == "Western"),
    "T-07 Partner 07": P07,
    "T-08 Jan+P07+Asian": (df["bulan"] == "2026-01") & P07 & (df["cuisine_type"] == "Asian"),
}

print(f"{'Test':<20} | {'Orders':<6} | {'Net (Rp)':<12} | {'Rev':<4} | {'Deliv':<5} | {'Late':<4}")
print("-" * 65)
for k, m in tests.items():
    n, net, rev, deliv, late = kpi(df[m])
    print(f"{k:<20} | {n:<6} | {net:<12,} | {rev:<4} | {deliv:<5} | {late:<4}")
```

```python
from pathlib import Path
import pandas as pd

base_dir = Path.cwd()
if base_dir.name == "notebooks":
    base_dir = base_dir.parent
csv_path = base_dir / "data" / "processed" / "capstone_swiftbite" / "swiftbite_capstone_analysis_ready.csv"
d = pd.read_csv(csv_path)

# --- A. Ulasan dan sentimen baru (Langkah 2) ---
r = d[d["review_text"].notna()].copy()
txt = r["review_text"].str.lower()
neg = r[r["sentiment_label"] == "Negative"]

print("Negative total:", len(neg))
print("Negative dengan rating 4-5:", int(neg["rating"].isin([4, 5]).sum()))
print("Positive dengan rating 1-2:", int(((r["sentiment_label"] == "Positive") & r["rating"].isin([1, 2])).sum()))

# Definisi resmi tema keluhan yang dikunci satu kali
tema = {
    "keterlambatan": r"terlambat|telat|\blambat|\blama",
    "dingin": r"dingin|tidak (?:terlalu )?hangat",
    "kemasan": r"kemasan|packaging|bocor|tumpah|rusak",
    "porsi sedikit": r"sedikit|kecil",
    "harga mahal": r"mahal",
    "driver/komunikasi": r"driver|kurir|komunikasi|sulit",
    "kesegaran/rasa": r"\bfresh|basi|hambar|\basin|\brasa",
}
nt = neg["review_text"].str.lower()
print("\nTema pada ulasan Negative (satu ulasan bisa masuk lebih dari 1 tema):")
for k, pat in tema.items():
    print(f"  {k:<20}: {int(nt.str.contains(pat, regex=True).sum())}")

# --- B. Analisis Sensitivitas Outlier O00084 & O00176 (Langkah 3) ---
print("\n=== ANALISIS SENSITIVITAS OUTLIER (LANGKAH 3) ===")
z_with = d.groupby("customer_zone")["net_payment_amount"].sum().sort_values(ascending=False)
d_no84 = d[d["order_id"] != "O00084"]
z_without = d_no84.groupby("customer_zone")["net_payment_amount"].sum().sort_values(ascending=False)
print("Peringkat Omzet Zona DENGAN O00084:")
for z, v in z_with.items():
    print(f"  {z:<8}: Rp {int(v):,}")
print("Peringkat Omzet Zona TANPA O00084 (Rp 1.749.000):")
for z, v in z_without.items():
    print(f"  {z:<8}: Rp {int(v):,}")

# Uji Outlier O00176 pada Molor West Zone
west_late = d[(d["customer_zone"] == "West") & (d["delivery_status"].str.lower() == "delivered") & (d["actual_delivery_minutes"] > d["promised_delivery_minutes"])]
west_delay_with = (west_late["actual_delivery_minutes"] - west_late["promised_delivery_minutes"]).mean()
west_late_no176 = west_late[west_late["order_id"] != "O00176"]
west_delay_no176 = (west_late_no176["actual_delivery_minutes"] - west_late_no176["promised_delivery_minutes"]).mean()
print(f"\nMolor West Zone DENGAN O00176 (n={len(west_late)}): +{west_delay_with:.2f} menit")
print(f"Molor West Zone TANPA O00176 (n={len(west_late_no176)}): +{west_delay_no176:.2f} menit")

# --- C. Uji Statistik West vs East (Langkah 5) ---
print("\n=== UJI SIGNIFIKANSI LATE RATE WEST VS EAST (LANGKAH 5) ===")
from scipy import stats
w_val = d[(d["customer_zone"] == "West") & (d["delivery_status"].str.lower() == "delivered")]
e_val = d[(d["customer_zone"] == "East") & (d["delivery_status"].str.lower() == "delivered")]
w_late = (w_val["actual_delivery_minutes"] > w_val["promised_delivery_minutes"]).sum()
e_late = (e_val["actual_delivery_minutes"] > e_val["promised_delivery_minutes"]).sum()
w_tot = len(w_val)
e_tot = len(e_val)

# Tabel Kontingensi 2x2
obs = [[w_late, w_tot - w_late], [e_late, e_tot - e_late]]
chi2, p_val, dof, ex = stats.chi2_contingency(obs, correction=False)
print(f"West: {w_late}/{w_tot} ({w_late/w_tot*100:.2f}%) vs East: {e_late}/{e_tot} ({e_late/e_tot*100:.2f}%)")
print(f"Chi-Square: {chi2:.4f} | p-value: {p_val:.4f} (p = {p_val:.2f} -> TIDAK SIGNIFIKAN)")
```
